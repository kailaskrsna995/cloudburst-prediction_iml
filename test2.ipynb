```python
from google.colab import drive
import pandas as pd

# Mount Google Drive
drive.mount('/content/drive')

# Define file path in Google Drive
file_path = "/content/drive/My Drive/cleaned_encoded.csv"
df=pd.read_csv(file_path)
```

```python
from sklearn.preprocessing import MinMaxScaler

# Separate features and target
X = df.drop('CloudBurstTomorrow', axis=1)
y = df['CloudBurstTomorrow']

# Initialize scaler
scaler = MinMaxScaler()

# Fit and transform features
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame (optional but useful)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
df
```

```python
X_scaled_df.describe()
```

```python
from sklearn.model_selection import train_test_split

# Use the scaled data and original labels
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, stratify=y, random_state=42)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping

# Define the ANN model
model1 = Sequential([
    Dense(64, activation='relu', input_shape=(X_train.shape[1],)),  # First hidden layer
    Dense(32, activation='relu'),                                   # Second hidden layer
    Dense(1, activation='sigmoid')                                  # Output layer
])

# Compile the model
model1.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Early stopping to avoid overfitting and restore best weights
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

# Train the model
history1 = model1.fit(
    X_train, y_train,
    epochs=75,                      # Give model room to improve
    batch_size=64,
    validation_split=0.2,

    verbose=1
)
```

```python
model_20 = Sequential([
    Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
model_20.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_20 = model_20.fit(
    X_train, y_train,
    epochs=20,
    batch_size=64,
    validation_split=0.2,

    verbose=1
)
```

```python
model_50 = Sequential([
    Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
model_50.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_50 = model_50.fit(
    X_train, y_train,
    epochs=50,
    batch_size=64,
    validation_split=0.2,

    verbose=1
)
```

```python
from sklearn.metrics import classification_report

# Model A (20 epochs) Evaluation
test_loss_20, test_acc_20 = model_20.evaluate(X_test, y_test, verbose=0)
y_pred_20 = (model_20.predict(X_test) > 0.5).astype(int)  # Convert predictions to binary
print(f"Model A (20 epochs) - Test Accuracy: {test_acc_20:.4f}")
print("Classification Report for Model A (20 epochs):")
print(classification_report(y_test, y_pred_20))
```

```python
# Model B (50 epochs) Evaluation
test_loss_50, test_acc_50 = model_50.evaluate(X_test, y_test, verbose=0)
y_pred_50 = (model_50.predict(X_test) > 0.5).astype(int)  # Convert predictions to binary
print(f"Model B (50 epochs[27]) - Test Accuracy: {test_acc_50:.4f}")
print("Classification Report for Model B (50 epochs):")
print(classification_report(y_test, y_pred_50))
```

```python
# Model C (75 epochs) Evaluation
test_loss_75, test_acc_75 = model1.evaluate(X_test, y_test, verbose=0)
y_pred_75 = (model1.predict(X_test) > 0.5).astype(int)  # Convert predictions to binary
print(f"Model C (75 epochs[23]) - Test Accuracy: {test_acc_75:.4f}")
print("Classification Report for Model C (75 epochs):")
print(classification_report(y_test, y_pred_75))
```

```python
# Separate the dataset into two parts: Class 0 and Class 1
class_0 = df[df['CloudBurstTomorrow'] == 0]
class_1 = df[df['CloudBurstTomorrow'] == 1]

# Calculate the number of rows to sample from Class 0 to achieve 70% Class 1
target_size = int(len(class_1) * 100 / 70) - len(class_1)

# Sample from Class 0 to add to the dataset to make it 70% Class 1
class_0_sample = class_0.sample(n=target_size, random_state=42)

# Combine the sampled data to create the new subset
subset_df = pd.concat([class_1, class_0_sample])

# Shuffle the dataset to ensure randomness
subset_df = subset_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Verify the class distribution
print(subset_df['CloudBurstTomorrow'].value_counts())
```

```python
# Features and target for the subset
X_subset = subset_df.drop(['CloudBurstTomorrow'], axis=1)
y_subset = subset_df['CloudBurstTomorrow']

# Scale the features
X_subset_scaled = scaler.transform(X_subset)  # Use the same scaler to scale
```

```python
subset_df
```

```python
from tensorflow.keras.callbacks import EarlyStopping

# Define early stopping to avoid overfitting
early_stop = EarlyStopping(
    monitor='val_loss',  # Monitor validation loss
    patience=5,          # Number of epochs to wait for improvement
    restore_best_weights=True  # Restore weights from the epoch with the best validation loss
)

# Define the secondary model (same architecture as Model B)
model2 = Sequential([
    Dense(64, activation='relu', input_shape=(X_subset_scaled.shape[1],)),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

# Compile the model
model2.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train the secondary model on the new subset with early stopping
history2 = model2.fit(
    X_subset_scaled, y_subset,
    epochs=50,  # You can tweak this as needed
    batch_size=64,
    validation_split=0.2,
    callbacks=[early_stop],  # Add early stopping callback
    verbose=1
)
```

```python
# Evaluate Secondary Model (trained on the subset with 70% ones)
print("Secondary Model Evaluation on Subset Dataset:")
model2_test_loss, model2_test_accuracy = model2.evaluate(X_subset_scaled, y_subset)
print(f"Test Accuracy on Subset: {model2_test_accuracy}")

# Predict on the subset test data
y_pred_subset = (model2.predict(X_subset_scaled) > 0.5).astype(int)

# Classification Report for precision, recall, f1-score on the subset dataset
from sklearn.metrics import classification_report
print("Classification Report for Secondary Model on Subset:")
print(classification_report(y_subset, y_pred_subset))
```

```python
import numpy as np
from sklearn.metrics import classification_report, accuracy_score

# Step 1: Predict with model_50 on the test set
model_50_predictions = model_50.predict(X_test)

# Ensure predictions are of shape (n_samples,) for comparison
model_50_predictions = model_50_predictions.flatten()

# Step 2: Identify false positives (model_50 predicts 1, but true label is 0)
false_positives = (model_50_predictions > 0.5) & (y_test.values == 0)  # Make sure y_test is in the correct format

# Step 3: Run those false positives through model2
model2_predictions = model2.predict(X_test[false_positives])

# Step 4: Replace model_50's incorrect predictions with model2's predictions
model_50_predictions[false_positives] = model2_predictions.flatten()

# Step 5: Evaluate the updated model (with model2 corrections)
updated_accuracy = accuracy_score(y_test, (model_50_predictions > 0.5))
print(f"Updated Test Accuracy: {updated_accuracy}")

# Print classification report for the updated predictions
print("Updated Classification Report:")
print(classification_report(y_test, (model_50_predictions > 0.5)))
```
